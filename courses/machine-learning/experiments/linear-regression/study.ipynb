# 一元线性回归实验

只使用 Python 标准库，用梯度下降拟合 y = 2x + 1。

在 VS Code / Jupyter 中选择 Python 内核后运行。计算使用该内核所在机器的资源。

```python
x = [-2.0, -1.0, 0.0, 1.0, 2.0]
y = [2 * value + 1 for value in x]
w, b = 0.0, 0.0
learning_rate = 0.05
```

## 训练

损失采用均方误差，每次更新前同时计算两个参数的梯度。

```python
for step in range(2000):
    errors = [w * xi + b - yi for xi, yi in zip(x, y)]
    loss = sum(error**2 for error in errors) / len(x)
    dw = 2 * sum(error * xi for error, xi in zip(errors, x)) / len(x)
    db = 2 * sum(errors) / len(x)
    w -= learning_rate * dw
    b -= learning_rate * db
    if step % 200 == 0:
        print(f"step={step:4d}  mse={loss:.8f}")
```

```python
print(f"w={w:.6f}, b={b:.6f}")
print(f"x=3 时预测 y={w * 3 + b:.6f}")
```

## 我的观察

- 把学习率改大或改小，损失如何变化？
